```python
import seaborn as sns
import pandas as pd
# Load the 'taxis' dataset
df = sns.load_dataset("taxis")
print(df.head())
print(df)
print(df.shape)
print(df.columns)
```

```python
import numpy as np


miss_columns = df.columns[df.isnull().any()]

num_cols = df.select_dtypes(include=np.number).columns

cat_cols = df.select_dtypes(include="object").columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

for col in cat_cols:
    if df[col].isnull().any() and not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])

df = df.dropna(subset=["pickup"])


print(df.isnull().sum())
```

```python
#Plot a line chart to visualize fare over time using pickup timestamp on the X-axis and fare on the Y-axis.
import matplotlib.pyplot as plt

df = df.sort_values("pickup")


plt.figure(figsize=(8, 5))

plt.plot(df["pickup"], df["fare"])

plt.title("Fare Over Time")
plt.xlabel("Pickup Timestamp")
plt.ylabel("Average Fare")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# The total fare for each pickup_borough.

borough_fare = df.groupby("pickup_borough")["fare"].sum()


plt.figure(figsize=(8, 5))

borough_fare.plot(kind="bar")

plt.title("Total Fare by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Total Fare")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# The distribution of trips based on payment method.

payment_counts = df["payment"].value_counts()

plt.figure(figsize=(5, 4))

plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of Trips by Payment Method")
plt.show()
```

```python
# The distribution of taxi trip distances.

plt.figure(figsize=(5, 3))

plt.hist(df["distance"], bins=20, edgecolor="black")

plt.title("Distribution of Taxi Trip Distance")
plt.xlabel("Distance")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# The distribution of tip amounts for each pickup_borough.

plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="pickup_borough",
    y="tip"
)

plt.title("Tip Distribution by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Tip Amount")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Display the number of trips in each pickup_borough

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="pickup_borough"
)

plt.title("Number of Trips by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Number of Trips")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# The relationship between distance and fare. Color the points based on pickup_borough.

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="distance",
    y="fare",
    hue="pickup_borough"
)

plt.title("Relationship Between Distance and Fare")
plt.xlabel("Distance")
plt.ylabel("Fare")

plt.tight_layout()
plt.show()
```

```python
# Show the correlation between numerical variables such as distance, fare, tip, tolls and total.

numerical_data = df[
    ["distance", "fare", "tip", "tolls", "total"]
]


correlation = numerical_data.corr()


plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# Visualize pairwise relationships between distance, fare, tip and total. Use pickup_zone to distinguish the data

pair_data = df[
    ["distance", "fare", "tip", "total", "pickup_zone"]
].dropna()


top_zones = pair_data["pickup_zone"].value_counts().head(5).index

pair_data = pair_data[
    pair_data["pickup_zone"].isin(top_zones)
]

sns.pairplot(
    pair_data,
    vars=["distance", "fare", "tip", "total"],
    hue="pickup_zone"
)

plt.show()
```

```python
# the distribution of fare for each payment method.

plt.figure(figsize=(9, 5))

sns.violinplot(
    data=df,
    x="payment",
    y="fare"
)

plt.title("Fare Distribution by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Fare")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
